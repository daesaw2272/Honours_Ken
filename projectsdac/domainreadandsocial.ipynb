```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("sdac2018_candidate_variables_raw.csv")
```

```python
need_vars = [
    "WHNASMOB",
    "WHNASSCR",
    "WHNASCOM",
    "NASSKINC",
    "WHNASEMO",
    "WHNASHOM",
    "WHNASPRO",
    "WNASMEAL",
    "WHNASFIN",
    "NASTRANS",
    "NAHCFCAR"
]

df = df[df[need_vars].eq(1).any(axis=1)].copy()

print(len(df))
```

```python
soc_away_activity_cols = [
    "SOC3AWYA",
    "SOC3AWYB",
    "SOC3AWYC",
    "SOC3AWYD",
    "SOC3AWYE",
    "SOC3AWYF",
    "SOC3AWYG",
    "SOC3AWYH",
    "SOC3AWYI",
    "SOC3AWYJ",
    "SOC3AWYK"
]
soc_home_activity_cols = [
    "SOCL3MA",
    "SOCL3MB",
    "SOCL3MC",
    "SOCL3MD",
    "SOCL3ME",
    "SOCL3MF",
]
```

```python
#convert them to binary
for col in soc_away_activity_cols:
    df[col] = (df[col] != 0).astype(int)

for col in soc_home_activity_cols:
    df[col] = (df[col] != 0).astype(int)
```

```python
for col in soc_away_activity_cols + soc_home_activity_cols:
    print(col, df[col].value_counts().to_dict())
```

```python
df["SOCAWAYCOUNT"] = (
    df[soc_away_activity_cols]
    .sum(axis=1)
    .astype(int)
)

df["SOCHOMECOUNT"] = (
    df[soc_home_activity_cols]
    .sum(axis=1)
    .astype(int)
)
```

```python
print(df["SOCAWAYCOUNT"].value_counts().sort_index())
print()

print(df["SOCHOMECOUNT"].value_counts().sort_index())
```

```python
df["SOCPARTICIPATIONTYPE"] = (
    (df["SOCAWAYCOUNT"] > 0).astype(int) +
    (df["SOCHOMECOUNT"] > 0).astype(int)
)
```

```python
print(df.columns.tolist())
```

```python
duplicate_wpm_cols = [
    col for col in df.columns
    if col.startswith("WPM") and col.endswith(".1")
]

print("Duplicate WPM columns:", len(duplicate_wpm_cols))
df = df.drop(columns=duplicate_wpm_cols)
```
