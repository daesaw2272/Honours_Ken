```python
import pandas as pd
```

```python
df = pd.read_csv("sdac2018processed.csv")
```

```python
need_vars = [
    "WHNASMOB",
    "WHNASSCR",
    "WHNASCOM",
    "NASSKINC",
    "WHNASEMO",
    "WHNASHOM",
    "WHNASPRO",
    "WNASMEAL",
    "WHNASFIN",
    "NASTRANS",
]

df = df[df[need_vars].eq(1).any(axis=1)].copy()

print(len(df))
```

```python
print("shape", df.shape)
print("rows", df.shape[0])
print("columns", df.shape[1])
print("duplicate", df.duplicated().sum())
```

```python
missing = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percent": (df.isna().mean() * 100).round(2)
})

missing = missing.sort_values(
    "Missing_Percent",
    ascending=False
)

print(missing)
```

```python
predictor_vars = [
    "AGEPC",
    "SEX",
    "EDLVLATB",
    "INCDECHD",
    "ARIAC",
    "SPENGPRO",
    "HHNOPSNB",
    "SOCAWAYCOUNT",
    "SOCHOMECOUNT",
    "SOCPARTICIPATIONTYPE",
    "SOCMORCO",
    "K10CATEG",
    "DISBSTAT",
    "SFHEALTH"
]

for col in predictor_vars:

    counts = df[col].value_counts(
        dropna=False
    ).sort_index()

    percentages = (
        df[col]
        .value_counts(dropna=False, normalize=True)
        .sort_index()
        * 100
    ).round(2)

    summary = pd.DataFrame({
        "Count": counts,
        "Percentage": percentages
    })

    print(summary)
```

```python
def frequency_table(df, col):
    counts = df[col].value_counts(dropna=False).sort_index()

    percentages = (
        df[col]
        .value_counts(dropna=False, normalize=True)
        .sort_index()
        * 100
    ).round(2)

    return pd.DataFrame({
        "Code": counts.index,
        "Count": counts.values,
        "Percentage": percentages.values
    })
```

```python
domain_mapping = {
    "Mobility": ("WHNASMOB", "RASEXMOB"),
    "Self-care": ("WHNASSCR", "RASEXSC"),
    "Communication": ("WHNASCOM", "RASEXCOM"),
    "Health care": ("NASSKINC", "RASEXHC"),
    "Cognition/emotion": ("WHNASEMO", "RASEXGUI"),
    "Household chores": ("WHNASHOM", "RASEXHOM"),
    "Property maintenance": ("WHNASPRO", "RASEXPRP"),
    "Meal preparation": ("WNASMEAL", "RASEXMEA"),
    "Reading/writing": ("WHNASFIN", "RASEXPAP"),
    "Transport": ("NASTRANS", "RASEXTRA")
}
```

```python
for domain, (need_var, outcome_var) in domain_mapping.items():

    domain_df = df[df[need_var] == 1]

    print(domain)
    print("Need variable:", need_var)
    print("Outcome variable:", outcome_var)
    print("People needing assistance:", len(domain_df))

    counts = (
        domain_df[outcome_var]
        .value_counts(dropna=False)
        .sort_index()
    )

    percentages = (
        domain_df[outcome_var]
        .value_counts(dropna=False, normalize=True)
        .sort_index()
        * 100
    ).round(2)

    print(pd.DataFrame({
        "Count": counts,
        "Percentage": percentages
    }))
```

```python
for domain, (need_var, outcome_var) in domain_mapping.items():

    error_1 = (
        (df[need_var] != 1) &
        (df[outcome_var] != 0)
    ).sum()

    error_2 = (
        (df[need_var] == 1) &
        (df[outcome_var] == 0)
    ).sum()

    print(f"\n{domain}")
    print("No need but RASEX != 0:", error_1)
    print("Need = 1 but RASEX = 0:", error_2)
```
